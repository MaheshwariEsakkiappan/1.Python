```python
# Create a function that checks whether the given number is Odd or Even 
class OddEven:
    def  findEvenOdd():
        num=int(input("Enter a number:"))
        if (num %2==0):
            print(num, "is Even number")
            message= str(num) +" is Even number"
        else:
            print(num, "is Odd Number")
            message= str(num) +" is Odd number"
        return message
            
```

```python
OddEven.findEvenOdd()
```

```python
#print area and perimeter of triangle using class and functions
class Triangle:
    def triangle():
        height=int(32) 
        breadth=int(34) 
        areaOfTriange= float((height*breadth)/2)
        print("Height :",height) 
        print("Breadth :",breadth) 
        print("Area formula: (Height*Breadth)/2 ")
        print("Area of Triangle:",  areaOfTriange)
        
        height1=int(2) 
        height2=int(4) 
        breadth1=int(4) 
        perimeterOfTriangle=height1+height2+breadth1 
        print("Height1 :",height1)
        print("Height2 :",height2)
        print("Breadth :",breadth1)
        print("Perimeter formula: Height1+Height2+Breadth ")
        print("Perimeter of Triangle:",  perimeterOfTriangle)
        return 
```

```python
Triangle.triangle()
```
