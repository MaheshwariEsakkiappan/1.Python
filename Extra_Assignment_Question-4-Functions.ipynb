```python
## Function definition
def marriageEligibility():
    gender=input("Your Gender:") 
    age=int(input("Your Age:")) 
    if((gender == "Male" or gender =="MALE" )and age>=21):
        eligibility="ELIGIBLE"
        print(eligibility)
    elif((gender == "Female" or gender == "FEMALE" )and age>=18):
         eligibility="ELIGIBLE"
         print(eligibility)
    else:
        eligibility="NOT ELIGIBLE"
        print(eligibility)
    return eligibility
```
