```python
 # Catagory the people by their age like children, adult, citizen, senior citizen...
age=int(input("Enter age of the person:"))
if(age<=18):
    print("Child")
elif(age <=45):
    print("Adult")
elif(age <=59):
    print("Citizen")    
else:
    print("Senior Citizen")
```
