```python
# A text-processing tool reverses a given word or sentence for formatting purposes.
#  Write logic to take a word or sentence as input and produce its reversed version.

input_statement="This is the reverse statement, Cool"
str= '' 
length=len(input_statement)
#print(length)
position=length-1
i=position
while (i>=0):
    str=str + input_statement[i]
    i=i-1
print("The reversed statement is : "+str)
```
