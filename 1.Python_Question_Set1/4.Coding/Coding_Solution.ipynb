```python
# 5.Write a Python program to check if a number is prime. Expected Output (for input 7):
# 7 is a prime number

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```
